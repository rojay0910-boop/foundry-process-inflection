```python
from pathlib import Path
import pandas as pd

# ========= 1) CSV files（path in computers） =========
INPUT_DIR = Path(
    r"C:\Users\rojay\OneDrive\ドキュメント\GitHub_portfolio\foundry-process-inflection\data"
)

# ========= 2) list of files to merge =========
FILES_TO_MERGE = [
    "process_nodes_TSMC.csv",
    "process_nodes_Samsung.csv",
    "process_nodes_Intel.csv",
   
]

# ========= 3) path to output (path to C:\Users\rojay\derived\） =========
OUTPUT_DIR = Path(r"C:\Users\rojay\derived")
OUTPUT_CSV = OUTPUT_DIR / "tableau_ready.csv"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("INPUT_DIR :", INPUT_DIR.resolve())
print("OUTPUT_CSV:", OUTPUT_CSV.resolve())
print("-" * 60)

# ========= 4) Check：if the directory and files exist =========
if not INPUT_DIR.exists():
    raise FileNotFoundError(f"No files found：{INPUT_DIR}")

paths = [INPUT_DIR / f for f in FILES_TO_MERGE]
missing_files = [p.name for p in paths if not p.exists()]
if missing_files:
    raise FileNotFoundError(f"File(s) not found. Please check the file name and directory.：{missing_files}")

print("Will merge these files:")
for p in paths:
    print(" -", p.name)
print("-" * 60)

# ========= 5) columns necessary (exit in each CSV） =========
REQUIRED_COLUMNS = [
    "company",
    "node_family",
    "technology_variant",
    "transistor_architecture",
    "euv_used",
    "first_volume_year",
    "application_focus",
    "production_status",
    "notes",
]

# ========= 6) Helper functions =========
def normalize_euv(x):
    if pd.isna(x):
        return ""
    s = str(x).strip().lower()
    if s in {"yes", "y", "true", "1"}:
        return "Yes"
    if s in {"no", "n", "false", "0"}:
        return "No"
    return str(x).strip()

def build_tech_label(row):
    nf = "" if pd.isna(row["node_family"]) else str(row["node_family"]).strip()
    tv = "" if pd.isna(row["technology_variant"]) else str(row["technology_variant"]).strip()
    if nf and tv:
        return f"{nf} ({tv})"
    return nf or tv or ""

# ========= 7) 讀取 + 欄位檢查 + 合併 =========
frames = []
for path in paths:
    df = pd.read_csv(path)

    missing_cols = [c for c in REQUIRED_COLUMNS if c not in df.columns]
    if missing_cols:
        raise ValueError(f"❌ {path.name} 缺少欄位：{missing_cols}")

    df["source_type"] = path.stem   # 例如 process_nodes_TSMC
    frames.append(df)

all_df = pd.concat(frames, ignore_index=True)

# ========= 8) Tableau-ready 欄位生成 =========
all_df["company"] = all_df["company"].astype(str).str.strip()
all_df["year"] = pd.to_numeric(all_df["first_volume_year"], errors="coerce")
all_df["euv_used"] = all_df["euv_used"].apply(normalize_euv)
all_df["technology_variant"] = all_df.apply(build_tech_label, axis=1)

# y_position（固定排序，Tableau 軌道穩定）
order = ["TSMC", "Samsung", "Intel"]
y_map = {c: i + 1 for i, c in enumerate(order)}
# 如果有其他公司沒在 order 裡，放到後面
unknown = [c for c in sorted(all_df["company"].unique()) if c not in y_map]
y_map.update({c: len(y_map) + i + 1 for i, c in enumerate(unknown)})
all_df["y_position"] = all_df["company"].map(y_map)

# technology_stage / power_delivery 先留空（你之後要做分類再補）
#all_df["technology_stage"] = ""
all_df["power_delivery"] = ""

# ========= 9) 輸出 schema（完全照你指定的欄位順序） =========
final_df = pd.DataFrame({
    "company": all_df["company"],
    "year": all_df["year"],
    "technology_stage": all_df["technology_stage"],
    "transistor_architecture": all_df["transistor_architecture"],
   # "power_delivery": all_df["power_delivery"],
    "technology_variant": all_df["technology_variant"],
    "euv_used": all_df["euv_used"],
    "source_type": all_df["source_type"],
    "y_position": all_df["y_position"],
    "notes": all_df["notes"],
}).drop_duplicates()

final_df.to_csv(OUTPUT_CSV, index=False, encoding="utf-8-sig")

print("🎉 DONE! CSV created:")
print("📄", OUTPUT_CSV.resolve())
print("📊 Rows:", len(final_df))

final_df.head()
```
